```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


data = np.array([
    [2, 0],
    [0, 2],
    [3, 3]
])


df = pd.DataFrame(data, columns=['X', 'Y'])
print("Original Data:")
print(df)

mean = np.mean(df, axis=0)
centered_data = df - mean
print("\nMean Centered Data:")
print(centered_data)

cov_matrix = np.cov(centered_data.T)
print("\nCovariance Matrix:")
print(cov_matrix)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("\nEigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)

sorted_indices = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[sorted_indices]
eigenvectors = eigenvectors[:, sorted_indices]

pc1 = eigenvectors[:, 0] 
projected_data = centered_data @ pc1  
print("\nProjected Data (1D along PC1):")
print(projected_data)
Point	X	Y
A	2	0
B	0	2
C	3	3

# Step 9: Visualize
plt.figure(figsize=(6, 6))
plt.scatter(centered_data['X'], centered_data['Y'], label='Original (Centered)', color='blue')
for i in range(len(projected_data)):
    proj_point = projected_data[i] * pc1
    plt.plot([0, proj_point[0]], [0, proj_point[1]], 'r--')
    plt.scatter(proj_point[0], proj_point[1], color='red')
    
origin = np.mean(centered_data, axis=0)
plt.quiver(*origin, *pc1, color='red', scale=3, label='PC1')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.title('PCA Projection onto First Principal Component')
plt.xlabel('X (Centered)')
plt.ylabel('Y (Centered)')
plt.legend()
plt.axis('equal')
plt.show()


```
